````python
import numpy as np
arr=np.random.randint(99,size=(7,5))
print(f"This is the original array\n\n{arr}\n\n And this would be the out put array:\n")

def mty(data):
    row,col=data.shape
    counter=0
    for i in range (col):
        for j in range (row):
            if data[j][i]%2!=0:
                counter+=1
    odd_arr=np.zeros(counter,dtype=int)
    a=0
    for m in range(col):
        for n in range (row):
            if data[n][m]%2!=0:
                odd_arr[a]=data[n][m]
                a+=1


    count=0
    for i in range (col):
        for j in range (row):
            if data[j][i]%2==0:
                count+=1
    even_arr=np.zeros(count,dtype=int)
    b=0
    for m in range(col):
        for n in range (row):
            if data[n][m]%2==0:
                even_arr[b]=data[n][m]
                b+=1

    print(f"Odd array: {odd_arr}\nEven array:{even_arr}")





mty(arr)
````
