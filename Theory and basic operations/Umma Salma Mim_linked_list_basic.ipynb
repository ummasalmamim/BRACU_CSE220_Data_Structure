```python
#creat link list
import numpy as np
arr=np.array([10,20,30,40,50,60,70])
print(arr)
class Node:
    def __init__(self,elem,next=None):
        self.elem=elem
        self.next=next
def creat (data):
    head=Node(data[0],None)
    tail=head
    for i in range(1,len(data)):
        test=Node(data[i],None)
        tail.next=test
        tail=test
creat(arr)
```

```python
#print link list
arr=[10,20,30,40,50,60,70]
class Node:
    def __init__(self,elem,next=None):
        self.elem=elem
        self.next=next
def creat (data):
    head=Node(data[0],None)
    tail=head
    for i in range(1,len(data)):
        test=Node(data[i],None)
        tail.next=test
        tail=test
    return head

def print_linklist(a):
    temp=a
    while temp!=None:
        print(temp.elem,end=" ")
        temp=temp.next

test1=creat(arr)
print_linklist(test1)
```
