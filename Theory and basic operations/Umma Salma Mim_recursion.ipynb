**Factorial**

```python
number=int(input("Enter Number:"))
def recusion_factorial(n):
    if n==1:
        return 1
    return n*recusion_factorial(n-1)
print (recusion_factorial(number) )
```

**Fibonacci numbers**

```python
number=int(input("Enter Number:"))
def fibonacci_numbers(n):
    if n==1 or n==0:
        return n
    return fibonacci_numbers(n-1)+fibonacci_numbers(n-2)
print(fibonacci_numbers(number))
```

**Summation of Normal Numbers**

```python
number=int(input("Enter Number:"))
def summation(n):
    if n==0:
        return 0
    return n+summation(n-1)
print(summation(number))
```

**Power**

```python
number=int(input("Enter Number:"))
pow=int(input("Enter Power:"))
def power (n,p):
    if p==1:
        return n
    return n*power(n,p-1)
print(power(number,pow))
```

**Determine the Len of link list**


```python
class Node:
    def __init__(self,elem,next=None):
        self.elem=elem
        self.next=next
```

```python
a=[10,20,30,40,50]
def create (arr):
    head=Node(arr[0])
    tail=head
    for i in range (1,len(arr)):
        temp=Node(arr[i],None)
        tail.next=temp
        tail=tail.next
    return head
t1=create(a)
```

```python
def count_link_list (data):
    if data.next==None:
        return 1
    return 1+count_link_list(data.next)
count_link_list(t1)
```
