````python
class University:
    def __init__(self,name,location,transport):
        self.u_name=name
        self.location=location
        self.transport=transport
````

````python
class Student(University):
    def __init__(self,name=None,id=None,course=None,cgpa=None,u_name=None,loc=None,trans=None):
        self.name=name
        self.id=id
        self.course=course
        self.__cgpa=cgpa
        University.__init__(self,u_name,loc,trans)
    def student_info(self):
        print("="*50)
        print(f"Student's Name: {self.name}\n ID: {self.id}\n Course Name: {self.course}")
        print(f"Student of: {self.u_name}\nLocation: {self.location}\nHas transport: {self.transport}")
        print("="*50)

    def setter(self,n):
        self.__cgpa=n
        print("CGPA Updated")
    def getter(self):
        print(self.__cgpa)
#========================Driver Code============================
s1=Student("FAISU",22103,"cse",3,"Bracu","Badda","Alif")
s1.student_info()
s1.name
s1.id
s1.setter(4)
s1.getter()

````
