```python
# Creating Node class
class Node:
    def __init__(self,element,next):
        self.element=element
        self.next=next
```

**Operations of Linked List**

Creation (converting array to a linked list)

```python
arr=np.array([10,20,30,40,50])
print(arr)

def creat_link_list(a):
    head=Node(a[0],None)
    tail=head
    for i in range(1,len(a)):
        n=Node(a[i],None)
        tail.next=n
        tail=tail.next
    return (head)
#creat_link_list(arr)
head=creat_link_list(arr) #we simply can right this
```

```python
def print_creat_link_list(data):
    temp=data
    while temp!=None:
        if temp.next!=None:
            print(temp.element,end="-->")
        else:
            print(temp.element)
        temp=temp.next #   or  temp= n
print_creat_link_list(head)
```

Count/length\
(Counting number of element in the link list)

```python
def count_element (data):
    temp=data
    counter=0
    while temp!=None:
        counter+=1             #  (In order to count the number of nodes present in link list we must need to
        temp=temp.next         #  count how many times the iteration process is running during a traversal.)
    return(counter)            # number of node = number of element
len_or_count=count_element(head)
print("Number of element in the link list: ", len_or_count)
```

NodeAt/ Index of\
(printing elements at a certain index)

```python
#i=int(input("Please enter index:" ))           #array:[10,20,30,40,50]
def nodeAt (data,idx):                         #index: 0  1  2  3  4
    temp=data
    index=0
    while temp!=None:
        if idx==index:
            return temp.element
        else :
            index+=1
            temp=temp.next
print(nodeAt(head,2))
```

```python
#i=int(input("Please enter index:" ))           #array:[10,20,30,40,50]
def objAt (data,idx):                         #index: 0  1  2  3  4
    temp=data
    index=0
    while temp!=None:
        if idx==index:
            return temp
        else :
            index+=1
            temp=temp.next
print(objAt(head,2))
```
