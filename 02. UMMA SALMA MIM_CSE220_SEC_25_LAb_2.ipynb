```python
#You must run this cell to print matrix and for the driver code to work
def print_matrix(m):
  row,col = m.shape
  for i in range(row):
    c = 1
    print('|', end='')
    for j in range(col):
      c += 1
      if(len(str(m[i][j])) == 1):
        print(' ',m[i][j], end = '  |')
        c += 6
      else:
        print(' ',m[i][j], end = ' |')
        c += 6
    print()
    print('-'*(c-col))

```

Task 1: Zigzag Walk

```python
def walk_zigzag(floor):
  row,column=floor.shape
  for a in range(0,row):
    if a%2==0:
        for b in range(0,column,2):
            print(floor[a][b],end=" ")
        print("")
    else:
         for c in range(column-1,-1,-2):
             print(floor[a][c],end=" ")
         print("")







floor = np.array([[ '3W' , '8B' , '4W' , '6B' , '1W' , '5B'],['3B' , '2W' , '1B' , '6W' , '3B' , '8W'],
                  ['9W' , '0B' , '7W' , '5B' , '3W' , '8B'],['2B' , '1W' , '3B' , '6W' , '0B' , '4W'],
                   ['1W' , '4B' , '2W' , '8B' , '6W' , '6B']])
print_matrix(floor)
print('Walking Sequence:')
walk_zigzag(floor)
#This should print
# 3W 4W 1W
# 8W 6W 2W
# 9W 7W 3W
# 4W 6W 1W
# 1W 2W 6W
```

Task 2: Landscape Screen

```python
def landscape(m):
    row,column=m.shape
    matrix=np.zeros((column,row),dtype=np.int64)
    for i in range(0,row,1):
        for j in range(0,column,1):
            matrix[j][i]=m[i][j]
    return (matrix)



m = np.random.randint(15,size=(4,3)) #This creates an np array of 4 x 3 where the values are random numbers from [0,15) range
print_matrix(m)
print()
print_matrix(landscape(m))
```

Task 3: Seating Arrangement

```python
import numpy as np
def arrangement(m, diff):
  row,column=m.shape
  for i in range(row):
    for j in range(column):
        if i==j:
            pass
        elif j>i:
            temp=m[i][j]-m[j][i]
            if temp<0:
                temp=-temp
            if temp!=diff:
                return False
  return True
#=======================================================================
m = np.array([[7,13,9,14],[12,8,15,11],[10,17,3,2],[15,10,1,4]])
print_matrix(m)
print()
returned_value = arrangement(m,1)
print(returned_value)#This should print False
unittest.output_test(returned_value, False)
print("===================================><===========================\n")
m = np.array([[8,15,7,12],[13,2,18,11],[9,20,5,2],[14,9,0,10]])
print_matrix(m)
print()
returned_value = arrangement(m,2)
print(returned_value)#This should print True
unittest.output_test(returned_value, True)
```

Task 4: Chess Piece

```python
def show_knight_move(knight):
    board=np.zeros((8,8),dtype=np.int64)
    row,column=knight
    if 0<=row<=7 and 0<=column<=7:
        board[row][column]=33
    else:
        return "Not Possible"

    if 0<=row-2<=7:
        if 0<=column-1<=7:
            board[row-2][column-1]=3
        if 0<=column+1<=7:
            board[row-2][column+1]=3
    else:
        pass


    if 0<=row+2<=7:
        if 0<=column-1<=7:
            board[row+2][column-1]=3
        if 0<=column+1<=7:
            board[row+2][column+1]=3
    else:
        pass


    if 0<=column-2<=7:
        if 0<=row-1<=7:
            board[row-1][column-2]=3
        if 0<=row+1<=7:
            board[row+1][column-2]=3
    else:
        pass


    if 0<=column+2<=7:
        if 0<=row-1<=7:
            board[row-1][column+2]=3
        if 0<=row+1<=7:
            board[row+1][column+2]=3
    else:
        pass



    return board


knight = (3,4)
chess_board = show_knight_move(knight)
print_matrix(chess_board)
```

```python
def topple(knight,rook):
    k_row,k_column=knight         #k= knight , r=rook
    r_row,r_column=rook
    board=np.zeros((8,8),dtype=np.int64)


    if 0<=k_row<=7 and 0<=k_column<=7:
        board[k_row][k_column]=33
    else:
        return "Not possible!"

    if 0<=r_row<=7 and 0<=r_column<=7:
        board[r_row][r_column]=5
    else:
        return "Not possible!"

    state="No one can kill  !"

    if 0<=k_row-2<=7:
        if 0<=k_column-1<=7:
            board[k_row-2][k_column-1]=3
            if board[k_row-2][k_column-1]==board[r_row][r_column]:
                state="Knight can kill !"

        if 0<=k_column+1<=7:
            board[k_row-2][k_column+1]=3
            if board[k_row-2][k_column+1]==board[r_row][r_column]:
                state="Knight can kill !"
    else:
        pass


    if 0<=k_row+2<=7:
        if 0<=k_column-1<=7:
            board[k_row+2][k_column-1]=3
            if board[k_row+2][k_column-1]==board[r_row][r_column]:
                state="Knight can kill !"
        if 0<=k_column+1<=7:
            board[k_row+2][k_column+1]=3
            if board[k_row+2][k_column+1]==board[r_row][r_column]:
                state="Knight can kill !"
    else:
        pass


    if 0<=k_column-2<=7:
        if 0<=k_row-1<=7:
            board[k_row-1][k_column-2]=3
            if board[k_row-1][k_column-2]==board[r_row][r_column]:
                state="Knight can kill !"
        if 0<=k_row+1<=7:
            board[k_row+1][k_column-2]=3
            if board[k_row+1][k_column-2]==board[r_row][r_column]:
                state="Knight can kill !"
    else:
        pass


    if 0<=k_column+2<=7:
        if 0<=k_row-1<=7:
            board[k_row-1][k_column+2]=3
            if board[k_row-1][k_column+2]==board[r_row][r_column]:
                state="Knight can kill !"
        if 0<=k_row+1<=7:
            board[k_row+1][k_column+2]=3
            if board[k_row+1][k_column+2]==board[r_row][r_column]:
                state="Knight can kill !"
    else:
        pass


    for i in range(0,8):
        if board[r_row][i]==board[k_row][k_column]:
            state="Rook can kill ! "
    for j in range(0,8):
        if board[j][r_column]==board[k_row][k_column]:
            state="Rook can kill !"

    return board,state


knight = (5,6)
rook = (5,1)
chess_board,ans= topple(knight,rook) #,ans
print_matrix(chess_board)
print(ans) #This should print Rook can kill
```
