```python
s=[10,20,30,40,50,60]
b=[1,0,0,1,0,1]
def playRight (sequence,beats):
    for i in range(0,len(beats),1):
        if beats[i]==1:
            temp=sequence[0]
            for j in range(1,len(sequence),1):
                sequence[j-1]=sequence[j]
            sequence[len(sequence)-1]=temp
    return(sequence)
print(playRight (s,b))
```

```python
def discardCards(cards,number):
    for i in range(0,len(cards),1):

        if cards[i]==number and cards[i+1]==number:
            shift(cards,i)
            shift(cards,i)

        elif cards[i]==number:
            shift(cards,i)

        else:
            pass
    return cards


def shift(data,limit):

    for j in range(limit+1,len(data)):
        data[j-1]=data[j]
    data[len(data)-1]=0

    return data
```
